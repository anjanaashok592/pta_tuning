```python
import glob
ng20_gwb = '/Users/ashokan/osu/sarahspta/data/NG_DSA_irn/'
pardir = ng20_gwb
timdir = ng20_gwb
pars = sorted(glob.glob(pardir+'*.par'))
tims = sorted(glob.glob(timdir+'*.tim'))
#check number of pulsars
print(len(pars))
print(len(tims))

#check first par and tim files
print(pars[0])
print(tims[0])
```

```python
#make enterprise pulsars
from enterprise.pulsar import Pulsar as ePulsar
ePsrs = []
for par,tim in zip(pars,tims):
    ePsr = ePulsar(par, tim,  ephem='DE440')
    ePsrs.append(ePsr)
    print('\rPSR {0} complete'.format(ePsr.name),end='',flush=True)
```

```python
#for the longest pulsar, plot the timespan of data with different backends
plt.figure(figsize=(20,8))
for ePsr in ePsrs:
    if ePsr.name == longest_pulsar.name:
        for backend in np.unique(ePsr.backend_flags):
            mask = ePsr.backend_flags == backend
            print('Backend:', backend, 'Timespan:', ePsr.toas[mask].max() - ePsr.toas[mask].min())
            t_span = ePsr.toas[mask].max() - ePsr.toas[mask].min()
            plt.plot(backend, t_span/(356*86400), 'o', color='blue')
plt.xlabel('Backend')
plt.ylabel('Timespan (years)')
plt.title('Timespan of Data with Different Backends for Longest Pulsar in the Dataset Faithful NG20 + a few more years of NG20-like + DSA')
plt.grid()
plt.show()
```

```python
#make a plot of the start and end times of the data for longest pulsar in the dataset with different backends
import astropy.time as atime

plt.figure(figsize=(20,8))
for ePsr in ePsrs:
    if ePsr.name == longest_pulsar.name:
        for backend in np.unique(ePsr.backend_flags):
            mask = ePsr.backend_flags == backend
            t_start_mjd = ePsr.toas[mask].min() / 86400
            t_end_mjd = ePsr.toas[mask].max() / 86400
            t_start = atime.Time(t_start_mjd, format='mjd').decimalyear
            t_end = atime.Time(t_end_mjd, format='mjd').decimalyear
            plt.plot([t_start, t_end], [backend, backend], color='blue', linewidth=5)
```
